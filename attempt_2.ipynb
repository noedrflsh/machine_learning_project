# Investigate outliers using the IQR

```python
outlier_results = []

for feature in continuous_or_discrete_numeric:
    feat = train_df[feature].dropna()
    Q1 = feat.quantile(0.25)
    Q3 = feat.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1-1.5*IQR
    upper_bound = Q3+1.5*IQR
    outlier_mask = (
        (feat < lower_bound) | (feat > upper_bound)
    )
    outlier_count = outlier_mask.sum()
    outlier_percentage = (
        outlier_count/len(feat) * 100
    )

    outlier_results.append({
        "feature": feature,
        "Q1": Q1,
        "Median": feat.median(),
        "Q3": Q3,
        "IQR":IQR,
        "lower_bound":lower_bound,
        "upper_bound":upper_bound,
        "outlier_count":outlier_count,
        "outlier_percentage":outlier_percentage
    })
outlier_summary = pd.DataFrame(outlier_results)
display(outlier_summary)
```

```python
delay_features = [
    "Departure Delay in Minutes",
    "Arrival Delay in Minutes"
]

for feature in delay_features:
    x = train_df[feature].dropna()

    q1 = x.quantile(0.25)
    median = x.quantile(0.50)
    q3 = x.quantile(0.75)

    iqr = q3 - q1

    lower_fence = q1 - 1.5 * iqr
    upper_fence = q3 + 1.5 * iqr

    outliers = x[
        (x < lower_fence) |
        (x > upper_fence)
    ]

    print(f"\nFeature: {feature}")
    print(f"Q1: {q1}")
    print(f"Median: {median}")
    print(f"Q3: {q3}")
    print(f"IQR: {iqr}")
    print(f"Lower fence: {lower_fence}")
    print(f"Upper fence: {upper_fence}")
    print(f"Potential outliers: {len(outliers)}")
    print(f"Outlier percentage: {len(outliers) / len(x) * 100:.2f}%")
```

```python
# Calculate correlations between rating features
rating_correlation = train_df[rating_columns].corr()

# Keep only the upper triangle to avoid duplicate pairs
upper_triangle = rating_correlation.where(
    np.triu(
        np.ones(rating_correlation.shape),
        k=1
    ).astype(bool)
)

# Convert the matrix into a list of feature pairs
correlation_pairs = (
    upper_triangle
    .stack()
    .reset_index()
)

# Give the columns meaningful names
correlation_pairs.columns = [
    "Feature 1",
    "Feature 2",
    "Correlation"
]

# Sort by absolute correlation, strongest first
correlation_pairs["Absolute correlation"] = (
    correlation_pairs["Correlation"].abs()
)

correlation_pairs = correlation_pairs.sort_values(
    by="Absolute correlation",
    ascending=False
)

display(correlation_pairs.head(10).round(3))
```

```python
# ============================================================
# 15. IQR-BASED OUTLIER ANALYSIS
# ============================================================

outlier_rows = []

for column in measurement_columns:

    values = train_df[column].dropna()

    if values.empty:
        continue

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_mask = (
        (values < lower_bound)
        | (values > upper_bound)
    )

    outlier_rows.append({
        "Feature": column,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Potential Outliers": int(outlier_mask.sum()),
        "Outlier Percentage": (
            outlier_mask.mean() * 100
        )
    })

outlier_summary = pd.DataFrame(outlier_rows)

display(outlier_summary.round(3))
```

```python
# ============================================================
# EXTREME OBSERVATIONS
# ============================================================

for column in measurement_columns:

    print(f"\nLargest values of {column}")

    display(
        train_df[column]
        .nlargest(10)
        .to_frame(column)
    )
```

```python
# ============================================================
# 16. MUTUAL INFORMATION FOR CLASSIFICATION (FIXED)
# ============================================================

from sklearn.feature_selection import mutual_info_classif
from sklearn.preprocessing import LabelEncoder

# Select features, excluding the target and identifier
mi_columns = [
    column for column in feature_columns
    if column not in id_columns
]

# Remove rows where the target is missing
mi_df = train_df[mi_columns + [TARGET]].dropna(
    subset=[TARGET]
).copy()

# Sample the dataset to reduce computation time
mi_df = mi_df.sample(
    n=min(50000, len(mi_df)),
    random_state=42
)

# Encode the target
target_encoder = LabelEncoder()
y_mi = target_encoder.fit_transform(
    mi_df[TARGET].astype(str)
)

X_mi = mi_df[mi_columns].copy()

# Identify categorical columns using their actual data types
categorical_mask = []

for column in X_mi.columns:

    is_categorical = (
        X_mi[column].dtype == "object"
        or isinstance(
            X_mi[column].dtype,
            pd.CategoricalDtype
        )
        or X_mi[column].dtype == "bool"
        or X_mi[column].dtype.name == "string"
    )

    if is_categorical:

        # Convert to object first so new string values are allowed
        X_mi[column] = (
            X_mi[column]
            .astype(object)
            .where(X_mi[column].notna(), "__MISSING__")
            .astype(str)
        )

        # Convert categories into integer codes
        X_mi[column] = LabelEncoder().fit_transform(
            X_mi[column]
        )

        categorical_mask.append(True)

    else:

        # Impute numerical missing values with the median
        X_mi[column] = X_mi[column].fillna(
            X_mi[column].median()
        )

        categorical_mask.append(False)

# Calculate mutual information
mi_scores = mutual_info_classif(
    X_mi,
    y_mi,
    discrete_features=categorical_mask,
    random_state=42
)

# Create a ranked summary
mi_summary = pd.DataFrame({
    "Feature": X_mi.columns,
    "Mutual Information": mi_scores
}).sort_values(
    "Mutual Information",
    ascending=False
)

display(mi_summary.round(4))
```

```python
# ============================================================
# STEP 3: LOGISTIC REGRESSION BASELINE
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

# Identify categorical and numerical features
categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool", "string"]
).columns.tolist()

numerical_features = [
    column for column in X_train.columns
    if column not in categorical_features
]

print("Categorical features:", categorical_features)
print("Numerical features:", numerical_features)

# Numerical preprocessing:
# 1. Replace missing values with the median.
# 2. Standardize numerical features.
numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing:
# 1. Replace missing values with the most frequent category.
# 2. Convert categories into one-hot encoded columns.
categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "onehot",
        OneHotEncoder(handle_unknown="ignore")
    )
])

# Apply the appropriate preprocessing to each feature group
preprocessor = ColumnTransformer([
    ("numerical", numerical_pipeline, numerical_features),
    ("categorical", categorical_pipeline, categorical_features)
])

# Combine preprocessing and Logistic Regression
baseline_model = Pipeline([
    ("preprocessor", preprocessor),
    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            random_state=42
        )
    )
])

# Train the model
baseline_model.fit(X_train, y_train)

# Predict probabilities for the positive class
positive_class_index = list(
    baseline_model.named_steps["classifier"].classes_
).index(True)

valid_probabilities = baseline_model.predict_proba(
    X_valid
)[:, positive_class_index]

# Evaluate ROC-AUC
baseline_auc = roc_auc_score(
    y_valid,
    valid_probabilities
)

print(f"\nBaseline validation ROC-AUC: {baseline_auc:.5f}")
```

```python
from catboost import CatBoostClassifier

TARGET = "satisfaction"
ID_COLUMN = "id"

# Use the original competition features.
feature_columns = [
    col for col in test_df.columns
    if col != ID_COLUMN
]

X_full = train_df[feature_columns].copy()
y_full = train_df[TARGET].copy()
X_test_final = test_df[feature_columns].copy()

# Identify categorical features.
cat_features_final = X_full.select_dtypes(
    include=["object", "category", "bool", "string"]
).columns.tolist()

# Convert categorical values to strings.
for col in cat_features_final:
    X_full[col] = X_full[col].fillna("__MISSING__").astype(str)
    X_test_final[col] = (
        X_test_final[col].fillna("__MISSING__").astype(str)
    )

# Train on the complete labeled dataset.
final_model = CatBoostClassifier(
    iterations=1375,
    depth=10,
    learning_rate=0.03,
    loss_function="Logloss",
    random_seed=42,
    verbose=100,
    allow_writing_files=False
)

final_model.fit(
    X_full,
    y_full,
    cat_features=cat_features_final
)

print("Final model trained!")
```
